# **Time Domain  Filters Math Toy Example**

<br>

### **The Signal We Want to Smooth**
FIR and IIR are just two different ways to do the same smoothing.

```
Signal: [3, 7, 2, 8, 1, 9, 4, 6]
```
<br>

## **FIR: Simple Moving Average (No Feedback)**

**The rule:** Replace each value with the average of it and its two neighbors.

The window is size 3. The weights are `[1/3, 1/3, 1/3]` — every sample in the window gets equal weight.

<br>

### Step 1: First Window

We need 3 samples to start. Take the first 3:

```
Signal: [3, 7, 2]  8, 1, 9, 4, 6
         └──┬──┘
         window

Average = (3 + 7 + 2) / 3 = 12 / 3 = 4.0

New signal: [4.0]
```

---

### Step 2: Slide Window Right

```
Signal: 3, [7, 2, 8]  1, 9, 4, 6
            └──┬──┘

Average = (7 + 2 + 8) / 3 = 17 / 3 = 5.67

New signal: [4.0, 5.67]
```

---

### Step 3: Slide Again

```
Signal: 3, 7, [2, 8, 1]  9, 4, 6
               └──┬──┘

Average = (2 + 8 + 1) / 3 = 11 / 3 = 3.67

New signal: [4.0, 5.67, 3.67]
```

---

### Step 4:

```
Signal: ..., 2, [8, 1, 9]  4, 6
                 └──┬──┘

Average = (8 + 1 + 9) / 3 = 18 / 3 = 6.0

New signal: [4.0, 5.67, 3.67, 6.0]
```

---

### Step 5:

```
Signal: ..., 8, [1, 9, 4]  6
                   └──┬──┘

Average = (1 + 9 + 4) / 3 = 14 / 3 = 4.67

New signal: [4.0, 5.67, 3.67, 6.0, 4.67]
```

---

### Step 6:

```
Signal: ..., 1, [9, 4, 6]
                    └──┬──┘

Average = (9 + 4 + 6) / 3 = 19 / 3 = 6.33

New signal: [4.0, 5.67, 3.67, 6.0, 4.67, 6.33]
```

---

### Final Comparison:

```
Original: [3,   7,   2,   8,   1,   9,   4,   6]
            ↘    ↘    ↘    ↘    ↘    ↘
Filtered:   [4.0, 5.67, 3.67, 6.0, 4.67, 6.33]
```

Look at what happened:
- The 9 (a spike) got pulled down to 6.33 by its neighbors 4 and 6.
- The 1 (a dip) got pulled up to 3.67 by 2 and 9.
- The filtered signal is smoother. The extreme jumps are reduced.

**This is an FIR filter.** Every output comes ONLY from the input samples in the window. No memory of past outputs.

<br>




## **IIR: Exponential Moving Average (With Feedback)**



 Each new output = 0.5 × current input + 0.5 × previous output.


<br>

### **The Signal We Want to Smooth**
FIR and IIR are just two different ways to do the same smoothing.

```
Signal: [3, 7, 2, 8, 1, 9, 4, 6]
```
<br>

### Step 1: First Sample

We have no previous output yet. We'll use 0 as the initial value.

```
Input[0] = 3
Previous output = 0 (we assume)

Output[0] = 0.5 × 3 + 0.5 × 0 = 1.5 + 0 = 1.5

New signal: [1.5]
```

---

### Step 2: Next Sample

```
Input[1] = 7
Previous output = 1.5

Output[1] = 0.5 × 7 + 0.5 × 1.5 = 3.5 + 0.75 = 4.25

New signal: [1.5, 4.25]
```

Notice: The input was 7, but the output is only 4.25. The old value (1.5) is pulling it down. The memory has influence.

---

### Step 3:

```
Input[2] = 2
Previous output = 4.25

Output[2] = 0.5 × 2 + 0.5 × 4.25 = 1.0 + 2.125 = 3.125

New signal: [1.5, 4.25, 3.125]
```

The input dropped to 2, but the output only dropped to 3.125. The memory of 4.25 is still pulling it up. Smoothing in action.

---

### Step 4:

```
Input[3] = 8
Previous = 3.125

Output[3] = 0.5 × 8 + 0.5 × 3.125 = 4.0 + 1.563 = 5.563

New signal: [1.5, 4.25, 3.125, 5.563]
```

---

### Step 5:

```
Input[4] = 1
Previous = 5.563

Output[4] = 0.5 × 1 + 0.5 × 5.563 = 0.5 + 2.781 = 3.281

New signal: [1.5, 4.25, 3.125, 5.563, 3.281]
```

---

### Step 6:

```
Input[5] = 9
Previous = 3.281

Output[5] = 0.5 × 9 + 0.5 × 3.281 = 4.5 + 1.641 = 6.141

New signal: [1.5, 4.25, 3.125, 5.563, 3.281, 6.141]
```

---

### Step 7:

```
Input[6] = 4
Previous = 6.141

Output[6] = 0.5 × 4 + 0.5 × 6.141 = 2.0 + 3.071 = 5.071

New signal: [1.5, 4.25, 3.125, 5.563, 3.281, 6.141, 5.071]
```

---

### Step 8:

```
Input[7] = 6
Previous = 5.071

Output[7] = 0.5 × 6 + 0.5 × 5.071 = 3.0 + 2.536 = 5.536

Final signal: [1.5, 4.25, 3.125, 5.563, 3.281, 6.141, 5.071, 5.536]
```

---

## Side by Side Comparison

```
Input:    [3,   7,   2,   8,    1,   9,    4,    6]

FIR (3pt avg):
          [4.0, 5.67, 3.67, 6.0,  4.67, 6.33]

IIR (50% feedback):
          [1.5, 4.25, 3.13, 5.56, 3.28, 6.14, 5.07, 5.54]
```

Both methods smooth the signal. Both reduce the spikes. But they work completely differently.

---




## Filters in the time domain

| | FIR (Moving Average) | IIR (Exponential Smoothing) |
|---|---|---|
| What you compute | Average of last N inputs | Mix of new input and old output |
| Memory | Finite (N samples) | Infinite (fades forever) |
| How many weights | N weights (window length) | 2 weights (input weight, feedback weight) |
| Recovery from spike | Gone after N samples | Fades exponentially |
| Feedback? | No | Yes |
| Stability | Always stable | Can diverge if feedback > 1 |
